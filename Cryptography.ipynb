```python
def caesar_encrypt(text, shift):
    result = ""
    for char in text:
        if char.isalpha():
            shift_base = 65 if char.isupper() else 97
            result += chr((ord(char) - shift_base + shift) % 26 + shift_base)
        else:
            result += char
    return result

def caesar_decrypt(text, shift):
    return caesar_encrypt(text, -shift)

# Example
print(caesar_encrypt("HELLO", 3))
```

```python
def vigenere_encrypt(text, key):
    result = ""
    key = key.lower()
    j = 0

    for char in text:
        if char.isalpha():
            shift = ord(key[j % len(key)]) - 97
            base = 65 if char.isupper() else 97
            result += chr((ord(char) - base + shift) % 26 + base)
            j += 1
        else:
            result += char
    return result

# Example
print(vigenere_encrypt("HELLO", "key"))
```

```python
def autokey_encrypt(plaintext, key):
    key = key + plaintext
    result = ""

    for i in range(len(plaintext)):
        p = ord(plaintext[i]) - 65
        k = ord(key[i]) - 65
        result += chr((p + k) % 26 + 65)

    return result

# Example
print(autokey_encrypt("HELLO", "KEY"))
```

```python
def rail_fence_encrypt(text, rails):
    fence = [[] for _ in range(rails)]
    row, step = 0, 1

    for char in text:
        fence[row].append(char)
        if row == 0:
            step = 1
        elif row == rails - 1:
            step = -1
        row += step

    return ''.join([''.join(r) for r in fence])

# Example
print(rail_fence_encrypt("HELLO WORLD", 3))
```

```python
import numpy as np

def hill_encrypt(text, key_matrix):
    text = text.replace(" ", "").upper()
    if len(text) % 2 != 0:
        text += "X"

    result = ""
    for i in range(0, len(text), 2):
        pair = [ord(text[i]) - 65, ord(text[i+1]) - 65]
        res = np.dot(key_matrix, pair) % 26
        result += chr(res[0] + 65) + chr(res[1] + 65)

    return result

key = [[3, 3], [2, 5]]
print(hill_encrypt("HELLO", key))
```

```python
from Crypto.Cipher import DES
from Crypto.Util.Padding import pad, unpad

key = b'8bytekey'
cipher = DES.new(key, DES.MODE_ECB)

plaintext = b'HELLO123'
ciphertext = cipher.encrypt(pad(plaintext, 8))

print("Encrypted:", ciphertext)

decrypted = unpad(cipher.decrypt(ciphertext), 8)
print("Decrypted:", decrypted)
```

```python
def rc4(key, text):
    S = list(range(256))
    j = 0
    out = []

    # KSA
    for i in range(256):
        j = (j + S[i] + key[i % len(key)]) % 256
        S[i], S[j] = S[j], S[i]

    # PRGA
    i = j = 0
    for char in text:
        i = (i + 1) % 256
        j = (j + S[i]) % 256
        S[i], S[j] = S[j], S[i]
        out.append(char ^ S[(S[i] + S[j]) % 256])

    return bytes(out)

# Example
key = b'key'
plaintext = b'hello'
cipher = rc4(key, plaintext)
print(cipher)
```

```python
from Crypto.Cipher import AES
from Crypto.Random import get_random_bytes
from Crypto.Util.Padding import pad, unpad

key = get_random_bytes(16)
cipher = AES.new(key, AES.MODE_CBC)

data = b"HELLO AES"
ciphertext = cipher.encrypt(pad(data, AES.block_size))

print("Encrypted:", ciphertext)

decipher = AES.new(key, AES.MODE_CBC, cipher.iv)
plaintext = unpad(decipher.decrypt(ciphertext), AES.block_size)

print("Decrypted:", plaintext)
```
